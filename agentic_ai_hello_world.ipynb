```python
import openai
from openai import OpenAI
from google.colab import userdata
openai_api_key = userdata.get("open_api_key")
#api_key = ""
```

```python
# Create a client (reads OPENAI_API_KEY from env)
client = OpenAI(api_key=openai_api_key)

SYSTEM_PROMPT = """
You are a helpful AI agent.
- Answer clearly and concisely.
- Ask for clarification only when truly needed.
"""

def run_agent():
    print("Simple AI agent. Type 'quit' to exit.\n")

    while True:
        user_input = input("You: ")
        if user_input.strip().lower() in {"quit", "exit"}:
            print("Agent: Goodbye! 👋")
            break

        # Call the LLM via Responses API
        response = client.responses.create(
            model="gpt-5-mini",   # or gpt-4.1-mini, etc.
            input=[
                {"role": "system", "content": SYSTEM_PROMPT},
                {"role": "user", "content": user_input},
            ],
        )

        # Convenient helper: output_text aggregates all text output
        agent_reply = response.output_text
        print(f"Agent: {agent_reply}\n")


if __name__ == "__main__":
    run_agent()

```

### Using Google's Gemini API as a free alternative

To use the Gemini API, you'll need an API key. If you don't already have one, create a key in [Google AI Studio](https://makersuite.google.com/key).

In Colab, add the key to the secrets manager under the "🔑" in the left panel. Give it the name `GOOGLE_API_KEY`. Then pass the key to the SDK:

```python
import google.generativeai as genai
from google.colab import userdata

# Fetch the API key from Colab's Secrets
GOOGLE_API_KEY = userdata.get('GOOGLE_API_KEY')

genai.configure(api_key=GOOGLE_API_KEY)
```

Now you can make API calls. For example, to generate a poem using the `gemini-pro` model:

```python
for m in genai.list_models():
  if 'generateContent' in m.supported_generation_methods:
    print(m.name)

# The output shows 'models/gemini-pro-latest' as an available model.
# Let's use that specific identifier for the GenerativeModel.
model = genai.GenerativeModel('models/gemini-pro-latest')

response = model.generate_content("Write a short, inspiring poem about mountains.")

print(response.text)
```
